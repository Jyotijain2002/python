```python
import sympy


```

```python
j= (3)**2
j_epsilon=(3+.001)**2
k=(j_epsilon-j)/0.001
print(f'J= {j}, J_epsilon= {j_epsilon}, dJ/dW ~= k = {k:.6f}')
```

```python
j= (3)**2
j_epsilon=(3+0.000000001)**2
k=(j_epsilon-j)/0.000000001
print(f'J= {j}, J_epsilon= {j_epsilon}, dJ/dW ~= k = {k:.6f}')
```

```python
J,W= sympy.symbols('J,W')
J=W**2
J
```

```python
dj_dw=sympy.diff(J)
dj_dw
```

```python
dj_dw.subs([(W,2)])
```

```python
dj_dw.subs([(W,3)])
```

```python
dj_dw.subs([(W,-8)])
```

```python
# J=2W
W, J=sympy.symbols('W,J')
J=2*W
J

```

```python
dj_dw=sympy.diff(J)
dj_dw
```

```python
dj_dw.subs([(W,-3)])
```

```python
J = 2*3
J_epsilon = 2*(3 + 0.001)
k = (J_epsilon - J)/0.001
print(f"J = {J}, J_epsilon = {J_epsilon}, dJ_dw ~= k = {k} ")
```
